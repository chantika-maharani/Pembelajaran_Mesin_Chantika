# 1.	Apa perbedaan supervised, unsupervised, dan reinforcement learning?
    Jawab :

Jadi supervised, unsupervised, sama reinforcement learning itu beda-beda cara belajarnya. Supervised itu kayak belajar pake kunci jawaban, datanya udah ada labelnya, jadi modelnya dibimbing buat nyari hubungan antara fitur sama jawaban yang bener, misalnya nebak harga rumah atau nentuin siswa lulus apa nggak. Unsupervised itu kebalikannya, datanya gak ada labelnya sama sekali, jadi modelnya harus nyari sendiri pola atau struktur yang tersembunyi di data itu, contohnya kayak ngelompokkin foto-foto tanpa dikasih tau kategorinya apa. Kalau reinforcement learning beda lagi, ini gak belajar dari data yang udah disiapin dari awal, tapi belajar lewat coba-coba sambil berinteraksi sama lingkungannya, dan tiap aksi yang diambil bakal dapet reward kalau bener atau punishment kalau salah, mirip kayak anak kecil yang dikasih reward pas rajin belajar dan dikasih punishment pas bolos les.
   

# 2.	Mengapa pembagian data latih dan uji penting?

      Jawab:
  Pembagian data latih sama data uji itu penting biar kita bisa tau model itu beneran paham pola datanya apa cuma hafalan doang. Soalnya kalau data yang dipake buat nguji sama persis kayak data yang dipake buat latihan, hasilnya pasti keliatan bagus terus, padahal itu bukan karena modelnya pinter, tapi karena dia udah "apal" sama data itu, istilahnya overfitting. Makanya data ujinya harus data yang belum pernah diliat sama modelnya, biar keliatan performa aslinya kalau nanti ketemu data baru di dunia nyata.

# 3.	Mengapa alur kerja ML disebut iteratif?

        Jawab:

  Alur kerja ML disebut iteratif karena prosesnya emang gak sekali jalan terus langsung kelar, tapi berulang-ulang. Jadi kalau abis dilatih terus dites ternyata hasilnya masih jelek atau kurang akurat, kita gak langsung nyerah, tapi balik lagi ke tahap sebelumnya buat dibenerin, entah itu dari datanya yang kurang, fiturnya yang kurang pas, atau modelnya yang perlu disetel ulang. Abis dibenerin, dilatih lagi, dites lagi, gitu terus muter sampe hasilnya dirasa udah cukup bagus. Makanya alurnya lebih kayak muter-muter (loop), bukan garis lurus dari awal sampe akhir.
