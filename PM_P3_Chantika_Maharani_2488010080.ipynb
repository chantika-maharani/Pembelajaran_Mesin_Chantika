## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns

df = sns.load_dataset('tips')
print(df.head())
df.shape
df.info()



```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
import seaborn as sns
import matplotlib.pyplot as plt

df_tips = sns.load_dataset('tips')
df.select_dtypes (include='number').hist(figsize=(9,6), bins=15)
plt.show()
sns.boxplot (data=df.select_dtypes (include='number'))
plt.title('Deteksi Outlier'); plt.show()
```

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
df_tips = sns.load_dataset('tips')
sns.scatterplot(data=df_tips, x='total_bill', y='tip', hue='time')
plt.show()
sns.pairplot (df, hue='time'); plt.show()

```

Yang dapat saya simpulkan adalah terdapat hubungan positif antara total_bill dan tip, di mana semakin besar tagihan maka semakin besar pula tip yang diberikan. Selain itu, transaksi pada waktu Dinner jumlahnya jauh lebih banyak dan cenderung memiliki total_bill yang lebih besar dibanding Lunch, sedangkan transaksi Lunch lebih sedikit dan nilainya cenderung lebih kecil



**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
num = df.select_dtypes(include='number')
sns.heatmap(num.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()

```

Pasangan fitur mana yang paling berkorelasi adalah tip vs total_bill
